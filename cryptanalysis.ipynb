## Was ist das überhaupt?
<div style="background: #f4f6f9; border-left: 4px solid #0056b3; padding: 10px;">
<b>Definition: Kryptoanalyse</b><br>
<i>"The techniques used by the adversary to attempt to "break" the cryptosystem." [Zitat nach Stinson]</i>
</div>

### Was ist ein Kryptosystem?
<div style="background: #f4f6f9; border-left: 4px solid #0056b3; padding: 10px;">
<b>Definition: Kryptosystem</b> als 5-Tupel $(\mathcal{P}, \mathcal{C}, \mathcal{K}, \mathcal{E}, \mathcal{D})$, bei dem folgende Eigenschaften gelten:
<ol>
  <li>$\mathcal{P}:$ endliche Menge aller möglichen <i>plaintexts</i></li>
  <li>$\mathcal{C}:$ endliche Menge aller möglichen <i>ciphertexts</i></li>
  <li>$\mathcal{K}:$ endliche Menge aller möglichen <i>keys</i></li>
  <li>$\forall K \in \mathcal{K:} \ \exists$ <i>encryption rule</i> $e_K \in \mathcal{E}$ $\land$ $\exists$ <i>(corresponding) decryption rule</i> $d_K \in \mathcal{D}$ mit Funktionen $e_K : \mathcal{P} \rightarrow \mathcal{C}$ und $d_K : \mathcal{C} \rightarrow \mathcal{P}$ $\implies$ $\forall x \in \mathcal{P} : d_K(e_K(x)) = x$ $\leftarrow$ die bedeutendste Eigenschaft</li>
</ol>
</div>

### Wie "bricht" / analysiert man es denn nun?
<div style="background: #f4f6f9; border-left: 4px solid #0056b3; padding: 10px;">
Dazu müssen wir erst einmal definieren, was der Öffentlichkeit bekannt ist<br>
$\rightarrow$ standardisiert unter dem <b>Kerckhoffs'schen Prinzip</b><br><br>

Inhaltlich gut fässt es m.M.n der Forscher und Professor (u.a. in Informationssicherheit) Steven M. Bellovin zusammen:<br>
<i>"design your system assuming that your opponents know it in detail [except the secret key]. (A former official at NSA's National Computer Security Center told me that the standard assumption there was that serial number 1 of any new device was delivered to the Kremlin.)"</i> $\implies$ Die Mengen $\mathcal{P}, \mathcal{C}, \mathcal{K}, \mathcal{E}, \mathcal{D}$ sowie der Algorithmus sind öffentlich bekannt, die Sicherheit beruht lediglich auf Geheimhalten des gewählten <i>secret key</i> $K \in \mathcal{K}$.
</div>
<br>
<div style="background: #f4f6f9; border-left: 4px solid #0056b3; padding: 10px;">
Um ein gutes Gewissen zu haben, dass unser Kryptosystem "sicher" ist, müssen wir so einen (evtl.) "Bruch" dieses Kryptosystems formal spezifieren, weshalb wir folgende drei Aspekte betrachten:

<hr style="border: none; border-top: 1px solid #ccc; margin: 15px 0;">

<b>1. Angriffsmodelle (<i>attack models</i>)</b><br>
Das Angriffsmodell spezifiziert die Informationsmenge, die dem Angreifer zur Verfügung steht. Folgende sind die vier üblichsten, bei denen stets ein <i> secret key </i> $K \in \mathcal{K}$ vorausgesetzt wird:
<ul>
  <li><b><i>known-ciphertext attack</i>:</b> Angreifer besitzt eine Menge von ciphertexten $\{C_1, \dots, C_m\} \subseteq \mathcal{C}$ mit $C_i = e_K(P_i)$</li>
  <li><b><i>known-plaintext attack</i>:</b> Angreifer besitzt eine Menge von Paaren aus plaintexten und korrespondierenden ciphertexten: $\{(P_1, C_1), \dots, (P_m, C_m)\} \subseteq \mathcal{P} \times \mathcal{C}$ mit $C_i = e_K(P_i)$</li>
  <li><b><i>chosen-plaintext attack</i>:</b> Angreifer kann gezielt plaintexte $P \in \mathcal{P}$ wählen und erhält ciphertexte $C = e_K(P) \in \mathcal{C}$</li>
  <li><b><i>chosen-ciphertext attack</i>:</b> Angreifer kann gezielt ciphertexte $C \in \mathcal{C}$ wählen und erhält plaintexte $P = d_K(C) \in \mathcal{P}$</li>
</ul>

<hr style="border: none; border-top: 1px solid #ccc; margin: 15px 0;">

<b>2. Angriffsziele (<i>adversarial goals</i>)</b><br>
Das Angriffsziel definiert den "Informationsleakgrad", ein paar Beispiele:
<ul>
  <li><b><i>complete break</i>:</b> Bestimmung des geheimen Schlüssels $K \in \mathcal{K}$ $\leftarrow$ stärkstes Ziel</li>
  <li><b><i>global deduction</i>:</b> Angreifer kann einen zuvor ungesehenen ciphertext entschlüsseln, ohne den Schlüssel zu kennen</li>
  <li><b><i>partial deduction</i>:</b> Angreifer kann aus einem zuvor ungesehenen ciphertext Teilinformationen gewinnen (z.B. bestimmte plaintext-Bits)</li>
  <li><b><i>distinguishability</i>:</b> Angreifer kann lediglich zwischen den Verschlüsselungen zweier gegebener plaintexte unterscheiden $\leftarrow$ schwächstes Ziel</li>
</ul>

<hr style="border: none; border-top: 1px solid #ccc; margin: 15px 0;">

<b>3. Sicherheitslevels (<i>security levels</i>)</b><br>
<ul>
  <li><b><i>computational security</i>:</b> angenommen es gäbe einen Algorithmus, der das Kryptosystem bricht; dann wäre dieser rechnerisch nicht durchführbar in vertretbarer Zeit mit aktuell, vefügbaren Rechnerressourcen</li>
  <li><b><i>provable / reductionist security</i>:</b> Brechen des Kryptosystems lässt sich mathematisch mittels Komplexitätstheorie auf ein gegebenes, <b>hartes</b> Problem reduzieren <i>(vgl. Primfaktorzerlegung, Diskreter Logarithmus aus Diskreter Mathe)</i></li>
  <li><b><i>unconditional security</i>:</b> Brechen des Kryptosystems selbst mit angenommen <b>"unendlicher"</b> Rechenleistung nicht möglich, da der dem Angreifer schlichtweg nicht genügend Informationen zur Verfügung stehen <i>(z.B. One-Time Pad</i>)</li>
</ul>
</div>

### Wie sehen solche Techniken konkret aus?
<div style="background: #f4f6f9; border-left: 4px solid #0056b3; padding: 10px;">
Um schließlich den theoretischen Bogen zur Eingangsdefinition zu schließen:<br><br>
Eine Technik, um das SPN "auszutricksen", nutzt die eben definierte <b>chosen-plaintext-attack</b>, mit dem Endziel eines <b>complete breaks.</b><br><br>
<b>$\rightarrow$ Differenzielle Kryptoanalyse</b>

<h2> Technik 1: Differenzielle Kryptoanalyse</h2>

<h3> Was ist das Problem (von SPN)?</h3>
<ul>
<i>Erinnerung: Fundamental basiert die Sicherheit eines SPN vollständig auf dessen nicht-lineare S-Boxen.</i><br>
$\rightarrow$ Problem: diese eigentlich tolle Nichtlinearität ist ein Jackpot für den Angreifer $\leftarrow$ sehr simple, grobe Abstraktion zum Einstieg<br>
</ul>
<br><br>
<i>Um diese Schwachstelle präziser verstehen zu können, definieren wir die Technik nach obigen Definitionen:</i>
<h4> Definition des Kryptosystems</h4> - verständnishalber erstmal runtergebrochen auf einzelne S-Box</h4>
<p>Aufgrund der Verwendung gegebener S-Box mit Blocklänge von $n=4$, definieren wir...</p>
<ul>
    <li>...die Mengen $\mathcal{P} = \mathcal{C} = \{0,1\}^4$</li>
    <li>...die Menge $\mathcal{K} = \{0,1\}^4$</li>
    <li>...in dieser Zeile nichts, aber die definierende Eigenschaft (4) gilt natürlich auch</li>
</ul>

<h4>Definition der Differenzen</h4>
<p>Seien $X', X'' \in \mathcal{P}$ zwei <em>chosen plaintexts</em> und $\pi_S: \{0,1\}^4 \rightarrow \{0,1\}^4$ die Substitutionsfunktion.</p>
<p>Die verwendeten Differenzen berechnen sich wie folgt:</p>
<ul>

$\rightarrow$ <strong>Eingabedifferenz:</strong> $\quad\Delta X = \underbrace{X' \oplus X''}_{\text{plaintexts}}$<br>
$\rightarrow$ <strong>Ausgabedifferenz:</strong> $\quad\Delta Y = \underbrace{\pi_S(X') \oplus \pi_S(X'')}_{\text{substituierte plaintexts}}$
</ul>

<div style="background: #f4f6f9; border-left: 4px solid #0056b3; padding: 15px; border-radius: 0 4px 4px 0; margin: 15px 0;">
    <strong>Anmerkung:</strong> Warum können wir die Eingabedifferenz der S-Box direkt anhand der plaintexte bestimmen, wenn es doch einen Key gibt?<br><br>
    $\rightarrow$ in einem SPN wird der plaintext $X$ zunächst mit dem first-round-subkey $K_1$ "gemixt" (vgl. <i>key mixing</i>), sodass $U = X \oplus K_1$ gilt, bevor die Daten in die S-Box gehen. Berechnet man nun diese tatsächliche Eingabedifferenz $\Delta U$, bemerkt man, dass der subkey keinen Effekt hat:
    <div style="text-align: center; padding: 10px 0; font-size: 1.1em;">
        $\Delta U = (X' \oplus K_1) \oplus (X'' \oplus K_1) = X' \oplus X'' \oplus (K_1 \oplus K_1) = X' \oplus X'' = \Delta X$
    </div>
    $\implies$ <strong>subkey kann bei der Betrachtung der Differenzen gekonnt ignoriert werden!</strong>
</div>

<h4>Definition der Propagationswahrscheinlichkeit</h4>
<p>Die <strong>Propagationswahrscheinlichkeit</strong> $P_D$, dass eine spezifische Eingabedifferenz $\Delta X$ zu einer Ausgabedifferenz $\Delta Y$ propagiert, ist der Quotient aus der Anzahl der erfüllenden plaintexte und $|\mathcal{P}| = 2^4 = 16$:</p>

<p>$$P(\Delta X \rightarrow \Delta Y) := \frac{|\{X' \in \{0,1\}^4 \mid \pi_S(X') \oplus \pi_S(X' \oplus \Delta X) = \Delta Y\}|}{16}$$</p>
<div style="background: #f4f6f9; border-left: 4px solid #0056b3; padding: 15px; border-radius: 0 4px 4px 0; margin: 15px 0;">
    <strong>Anmerkung:</strong> warum $\pi_S(X' \oplus \Delta X)$ und nicht $\pi_S(X'')$?<br><br>
    Um die Propagationswahrscheinlichkeit für eine <b>feste</b> Eingabedifferenz $\Delta X$ zu bestimmen, müssen wir alle möglichen Eingabepaare $(X', X'')$ betrachten, die diese Differenz erzeugen.<br> Da $\Delta X = X' \oplus X''$ gilt, ist die Wahl von $X'$ zwar frei, <b>aber</b> $X''$ durch $X'' = X' \oplus \Delta X$ zwingend festgelegt. <i>(o.B.d.A. geht auch vice versa, aber iwr machen des hier mit $X'$ als Wahl)</i>
    <br><br>
</div>
<p> Um nun auf das anfangs angesprochene Problem zurückzukommen: in einer idealen (d.h. absolut random) S-Box müsste gelten $$\forall P(\Delta X \rightarrow \Delta Y) = \frac{1}{2^n} = \frac{1}{16}$$<br>
<b>ABER:</b> so wie S-Box funktioniert, werden wir in der Realität Paare $\underbrace{(\Delta X, \Delta Y)}_{\text{Differential}}$ finden, für die gilt: $$P(\Delta X \rightarrow \Delta Y)\gg \frac{1}{16}$$
<div style="background: #f4f6f9; border-left: 4px solid #0056b3; padding: 15px; border-radius: 0 4px 4px 0; margin: 15px 0;">
    <strong>Anmerkung:</strong> wie entstehen diese Differentiale mit so hoher $P_D$ überhaupt?<br><br>
    Eine Idealverteilung kann auf Grund drei fundamentaler Eigenshcaften mathematisch nicht existieren:
    <ul style="margin-top: 5px; margin-bottom: 0;">
        1. <strong>Bijektivität:</strong> trivialerweise kommt bei $\Delta{X} = 0$ immer $\Delta{Y} = 0$ raus $\implies \forall \Delta{X} > 0: \Delta{Y} > 0$ <i>(anders gesagt: die Spalte $\Delta{Y} = 0$ ist irrelevant für diese Fälle)</i><br>
        2. <strong>XOR-Kommutativität:</strong> $X' \oplus X'' = X'' \oplus X' \implies \Delta(X'\oplus X'') = \Delta(X'' \oplus X') \implies$ haben gleiche Ausgabedifferenz $\Delta{Y} \implies$alle absoluten Propagationen müssen <b>gerade</b> sein<br>
        (3. <strong>Schubfachprinzip:</strong> die $16$ möglichen Eingabedifferenzen müssen nach <i>(1.)</i> auf $15$ mögliche Ausgabedifferenzen verteilt werden, zusätzlich treten diese nach <i>(2.)</i> nur in geraden Zahlen auf $\implies$ gewünschte Propagationen; ergibt sich aus (2) und (3))
    </ul>
</div>
<p>Der sogenannte <strong>Difference Distribution Table (DDT)</strong> fasst die Propagationen für alle Differentiale $(\Delta X, \Delta Y)$ zusammen.</p>

Erstmal importieren wir die in *modules/...* ausgelagerte Klasse: DifferentialAttack, um das demonstrieren zu können; wir instanziieren auch gleich ein Objekt.

<p>Zur Instanziierung ist auch gleich anpassbare, wie viele korrekte Paare <i>(right pairs)</i> $c$ man als Puffer haben will, um von zufälligen Rauschen erfolgreich unterscheiden zu können, daraus ergibt sich mit der Propagationswahrscheinlichkeit $P_D$ die ungefähre Anzahl benötiger Paare an <i>plaintexts</i> $N_D$ [Ist für die erfolgreiche Demonstration; Erklärung folgt später präziser, nur damit man grob den Sinn greifen kann]</i></p>

<p>$$N_D \approx \frac{c}{P_D}$$</p>
<br>
Bei Instanziierung werden zudem gleich die S-Box-Arrays initialisiert und die 16x16 DDT aufgestellt.
Wir wählen hier $c=38$, da das nach Stinson eine praktikable Wahl [genauere Erklärung folgt später]

```python
from modules import DifferentialAttack
attack = DifferentialAttack(expected_right_pairs=38)
```

<p>Um die Anfälligkeit der S-Box praktisch zu demonstrieren, legen wir als Eingabedifferenz $\Delta X =$ <b>0x0B</b>  und als seine korrespondierende Ausgabedifferenz $\Delta Y =$ <b>0x02</b> fest<br>
$\rightarrow$ indem wir uns die Propagationen hierfür anschauen, ist es uns möglich $P_D($0x0B$\rightarrow$ 0x02$)$ zu berechnen, schauen wir mal was passiert...</p>

```python
dx = 0x0b
dy = 0x02

propagations = attack.ddt[dx][dy]

print(f"In {propagations} von 16 Fällen kriegen wir bei der Eingabedifferenz diese Ausgabedifferenz.")
```

<p>$P_D($0x0B$\rightarrow$ 0x02$) = \frac{8}{16} = \frac{1}{2} = 50\%$ Propagationswahrscheinlichkeit, dass wir für dieses $\Delta{X}$ dieses $\Delta{Y}$ erhalten.</p><br>

<br>
<p>Anstatt das nun einzeln für alle $16^2 = 256$ Möglichkeiten (inkl. $\Delta X/Y= 0$, also eigentlich $256-16 = 240$) durchzuführen, benutzen wir die DDT, um so Differentiale zu entdecken. Wir haben sie als $16\times16$ Matrix implementiert.</p>

```python
ddt = attack.ddt

hex_diff_labels = [f"0x{idx:X}" for idx in range(16)]
table_data = [["ΔX↓ | ΔY→"] + hex_diff_labels]

for idx, row in enumerate(ddt):
    table_data.append([f"0x{idx:X}"] + row)

ddt_table = table(table_data, header_row=True, header_column=True, frame=True, align='center')
print(ddt_table)
```

Nun sieht man nochmal klar und deutlich die Schwäche der S-Boxen.<br>
Wir haben uns jetzt die Propagationswahrscheinlichkeit für <b>eine</b> isolierte S-Box angeschaut.<br>
<i>Erinnerung: unser zu "knackendes" System ist aber ein 4-Runden SPN, wie machen wir das nun am schlausten?</i>

<h3> Mehrrundenpropagation: 2 Möglichkeiten</h3>
<p>Um ein $R$-Runden SPN anzugreifen, muss auch die Wahrscheinlichkeit eines R-1-Runden-Differential bestimmt werden.<br>
Hierbei wird zwischen zwei Methoden unterschieden:</p>

<u>1. Die praktische Berechnung der $P_D$ vom Differential</u>

<div style="background: #f4f6f9; border-left: 4px solid #0056b3; padding: 15px; border-radius: 0 4px 4px 0; margin: 15px 0;">
    <strong>Definition: differential characteristic / (differential) Trail</strong><br><br>
    Ein Trail spezifiziert exakt <em>eine</em> feste Abfolge aktiver S-Boxen über mehrere Runden hinweg und ignoriert alle anderen möglichen Pfade. Er wird konstruiert, indem Differentiale aufeinanderfolgender Runden so verkettet werden, dass die permutierte Ausgabedifferenz der Runde $r$ exakt der Eingabedifferenz der S-Boxen in Runde $r+1$ entspricht.
</div>

<p><strong>Markov-Chiffre-Annahme:</strong></p>
<p>Um die Gesamtwahrscheinlichkeit $P_D$ dieses konstruierten Trails zu berechnen, nimmt man an, dass die verschiedenen Propagationswahrscheinlichkeiten der aktiven S-Boxen <b>voneinander unabhängige Ereignisse</b> sind. Obwohl diese Annahme evtl. nicht korrekt ist, ist sie praktisch.<br>
$\rightarrow$ sie erlaubt uns, $P_D$ als Produkt der einzelnen S-Box-Wahrscheinlichkeiten entsprechend zu berechnen:</p>

<p>$$P_D(T) = \prod_{i=1}^{\gamma_T} \beta_{i,T} \quad \text{(wobei } \gamma_T \text{ Anzahl aktiver S-Boxen im Trail } T \text{ und } \beta_{i,T} \text{ d. Wsl.)}$$</p>
<u>*2. Die exakte Berechnung der $P_D$ vom Differential</u><br>
Die theoretisch exakte Wahrscheinlichkeit eines <strong>Differentials</strong> $(\Delta X, \Delta Y)$ über $R-1$ Runden entspricht der Summe der Wahrscheinlichkeiten <b>aller</b> gültigen <i>differential characteristics (trails)</i> $\Omega$, die $\Delta X$ auf $\Delta Y$ abbilden.

Da die <strong>Unabhängigkeitsannahme</strong> weiterhin Anwendung findet, berechnet sich die Gesamtwahrscheinlichkeit durch die Summation dieser <i>trails</i>:

Sei $\Omega = \{ T \mid T \text{ ist ein gültiger Trail von } \Delta X \rightarrow \Delta Y \}:$$$P(\Delta X \rightarrow \Delta Y) = \sum_{T \in \Omega} P_D(T) = \underbrace{\sum_{T \in \Omega} \left( \prod_{i=1}^{\gamma_T} \beta_{i,T} \right)}_{\text{Einsetzen der Mgl. 1}}$$

<div style="background: #f4f6f9; border-left: 4px solid #0056b3; padding: 15px; border-radius: 0 4px 4px 0; margin: 15px 0;">
    <strong>Anmerkung:</strong><br><br>
    Diese Gleichung zeigt, dass die exakte Berechnung rechnerisch nicht vertretbar ist. Denn bei einer Blocklänge von $n$ existieren $2^n - 1$ mögliche <i>(nicht-triviale)</i> "Zwischen"differenzen.<br>
$\rightarrow$  bei einem $(R-1)$-Runden-Differential sind Start und Ziel fest $\implies$ es existieren $R-2$ Zwischenzustände <br>
$\Rightarrow$ Komplexität wächst $\in$ $\mathcal{O}((2^n - 1)^{R-2})$; lassen wir das mal lieber sein... und suchen uns dominante Trails mit hohen Wahrscheinlichkeit $P_D$
</div>

<h3> Anwendung: beste "R-1"-Runden Trails</h3>
<p>Für das 4-Runden SPN ($R=4$) werden nun gezielt Trails über 3 Runden ($R-1=3$) konstruiert.</p>
$\rightarrow$ die erwartete Ausgabedifferenz nach Runde 3 dient als Eingabedifferenz für die letzte Runde<br>
$\rightarrow$ ermöglicht den Angriff auf spezifische Bits des letzten subkeys ($K_5$)

```python
# anlegen des "r-1" = 3 runden-trail / characteristic
round_length= attack.num_rounds - 1
trails = attack.get_highly_probable_trails(round_num=round_length)

table_data = [["Rang", "Eingabedifferenz ΔX", "erwartete Ausgabedifferenz ΔY (nach Runde 3)", "Propagationswahrscheinlichkeit"]]

for idx, trail in enumerate(trails[:5]):
    pref, prob, in_diff, out_diff = trail

    exact_prob = Integer(round(prob * 4096)) / 4096

    table_data.append([
        f"#{idx+1}",
        f"0x{in_diff:04X}",
        f"0x{out_diff:04X}",
        f"{exact_prob} ≈ {prob:.6f}"
    ])

trails_table = table(table_data, header_row=True, frame=True, align='center')
print(trails_table)
```

<h3>Einschub: right und wrong Pairs, anfangs versprochene fortsetzende Erklärung</h3>

<p><b>1. right pairs</b><br>
Ein plaintextpaar, dessen tatsächliche interne Differenzen über <b>alle</b> Runden hinweg exakt unserem gewählten Trail folgen. Nur diese Paare erzeugen am Ende die erwartete Ausgabedifferenz $\Delta Y$ und geben dem <i>korrekten</i> subkey einen "Hit"</p>

<p><b>2. wrong pairs (sogenanntes Rauschen)</b><br>
Ein Paar, das vom vorhergesagten Pfad abweicht. Diese Paare erzeugen völlig zufällige Ausgabedifferenzen. Sie verteilen ihre "Hits" gleichmäßig und zufällig auf <i>falsche</i> Schlüsselkandidaten.</p>

<p><u>Puffer</u> $c$<br>
Um den richtigen Schlüssel einigermaßen verläslich aus dem Rauschen der Wrong Pairs herauszufiltern, müssen wir eine Mindestanzahl an Right Pairs als Puffer festlegen.<br>
Diese Konstante $c$ haben wir bei der Instanziierung auf $c=38$ gesetzt $\implies >99\%$ Wahrscheinlichkeit, korrekte subkeys vom Rauschen zu unterscheiden (nach stinson)<br>
Daraus lässt sich nun in etwa die Menge der insgesamt benötigten plaintextpaare $N_D$ für den Angriff berechnen.</p><br>
<i>Erinnerung:</i> $$N_D \approx \frac{c}{P_D}$$


```python
best_trail = trails[0]
prob_best_trail = best_trail[1]
expected_c = attack.expected_right_pairs

# nach obiger formel
n_d = int(expected_c / prob_best_trail)

table_data = [
    ["Parameter", "Wert"],
    ["Trailwahrscheinlichkeit [P_D(T)]", f"{prob_best_trail:.6f}"],
    ["Ziel-Puffer an Right Pairs [c]", f"{expected_c}"],
    ["Benötigte plaintextpaare [N_D]", f"{n_d}"]
]

params_table = table(table_data, header_row=True, frame=True, align='center')

print("Unsere Angriffsparameter für die key extraction lauten:")
print(params_table)
```

<h3> partial decryption & key extraction</h3>

<p>Mit unserem gewählten Trail fokussieren wir uns nun auf die Extraktion des sogenannten <strong>target partial subkey</strong>.</p>
<p>$\rightarrow$ Suchraum beschränkt sich auf die Bits des letzten subkeys $K_5$, welche mit den Ausgängen der durch unseren Trail aktiven S-Boxen der finalen Runde verknüpft sind</p>
<p>$\rightarrow$ Angriff erfordert eine partial decryption der letzten Runde für alle potenziellen Werte des Suchraum von $K_5$</p>
<div style="background: #f4f6f9; border-left: 4px solid #0056b3; padding: 15px; border-radius: 0 4px 4px 0; margin: 15px 0;">
    <strong>Exkurs: Was ist ein encryption oracle?</strong><br><br>
    Der Begriff stammt aus der Berechenbarkeitstheorie (Alan Turing, 1939). In der Kryptoanalyse abstrahiert das <i>Encryption Oracle</i> das System des Opfers als absolute Blackbox: Der Angreifer erhält für beliebig gewählte plaintexte die korrekten ciphertexte, ohne den geheimen Schlüssel zu kennen.<br><br>
    $\implies$ formalisiert die exakte Angreiferfähigkeit einer chosen-plaintext-attack
</div>
<p>Seien $(C', C'')$ die vom encryption oracle entstandenen ciphertextpaare. Für jeden potenziellen subkey-Kandidaten $K_5^*$ wird der Zustand <i>vor</i> der letzten S-Box ($U_4$) isoliert partiell entschlüsselt:</p>

<p>$$V_4 = C \oplus K_5^* \quad \implies \quad U_4 = \pi_S^{-1}(V_4)$$</p>

<p>Stimmt die berechnete Differenz exakt mit der erwarteten Ausgabedifferenz $\Delta Y$ unseres 3-Runden-Trails überein?</p>

<p>$$\Delta U_4 = U_4' \oplus U_4'' \stackrel{?}{=} \Delta Y$$</p>

<p>Falls ja, inkrementiert der "hit-count" für $K_5^*$<br>
$\rightarrow$ aufgrund des ohnehin dominanten "hit-counts" der <i>right pairs</i> (und anhand unseres festgelegten Puffers $c$) erhalten wir statistisch den korrekten <strong>target partial subkey</strong>.</p>

<h3>Vom 3-Runden-Trail zum Complete Break in zwei Phasen</h3>
<p>Nachdem wir nun die Theorie zu <em>differential trails</em> abgeklappert haben, führen wir nun den <i>complete break</i> in zwei Phasen durch.</p>

<div style="background: #f4f6f9; border-left: 4px solid #0056b3; padding: 15px; border-radius: 0 4px 4px 0; margin: 15px 0;">
    <strong>Phase 1:</strong> iterative partial decryption ($K_5 \rightarrow K_4 \rightarrow K_3$)</strong><br>
    Wir iterieren rückwärts, beginnend mit $K_5$ und extrahieren jeweils die <i>round/subkeys</i>:<br>
    <ul style="margin-top: 5px; margin-bottom: 0;">
$\rightarrow$ <strong>Runde 5 ($K_5$):</strong> Auf alle potentiellen Kandidaten des <i>subkey</i> wird eine partial decryption angewandt <i>(mittels 3-Runden-Trail)</i><br>
$\rightarrow$ <strong>Runde 4 ($K_4$):</strong> $K_5$ bekannt, partial decryption der ciphertexte <i>(mit 2-Runden-Trail auf neuen Endzustand)</i><br>
$\rightarrow$ <strong>Runde 3 ($K_3$):</strong> $K_5, K_4$ bekannt, partial decryption der ciphertexte <i>(mit abschließendem 1-Runden-Trail)</i><br>
    </ul>
</div>

<div style="background: #f4f6f9; border-left: 4px solid #28a745; padding: 15px; border-radius: 0 4px 4px 0; margin: 15px 0;">
    <strong>Phase 2:</strong> bisschen Algebra ($K_2$ & $K_1$)<br>
    <i>Anmerkung: Diese algorithmische Optimierung zur deterministischen Eliminierung von $K_2$ stammt <b>NICHT</b> aus heys/stinson, sondern wurde aus entsprechend verlinkten py-skript abgeleitet.</i><br><br>
    <i>differential trails</i> werden für sehr kurze Distanzen ineffizient
     $\rightarrow$ stattdessen: nutzen wir die bekannten subkey $K_5, K_4, K_3$ für eine Rückrechnung
</div>

<h4> *Beweis zur Optimierung in Phase 2 (orientiert an heys Nomenklatur)</h4>
<strong>Voraussetzung bzw. Definitionen:</strong><br>
$\rightarrow$ $U_r:=$ Zustand <i>vor</i> S-Box in Runde $r$ (nach key-mixing)<br>
$\rightarrow$ $W_r:=$ Zustand <i>nach</i> P-Box in Runde $r$<br>
$\rightarrow$ $K_r:=$ Subkey der Runde $r$.<br>
$\rightarrow$ $U_2 = W_1 \oplus K_2:=$ key-mixing-"operator"<br>
$\rightarrow$ $X_{r,i}:=$ Zustand $X$ in Runde $r$ für $i$-tes Element des Paares

<strong>Behauptung:</strong>
<ul style="margin-top: 5px; margin-bottom: 15px;">
    Die Differenz $\Delta U_2 = U_{2,1} \oplus U_{2,2}$ eliminiert $K_2$ vollständig und liefert exakt die Ausgabedifferenz der ersten Runde:
    $$\Delta U_2 = \Delta W_1$$
</ul>

<strong>Beweis:</strong>
<ul style="margin-top: 5px; margin-bottom: 0;">
    $$\Delta U_2 = U_{2,1} \oplus U_{2,2} \iff
    \Delta U_2 = (W_{1,1} \oplus K_2) \oplus (W_{1,2} \oplus K_2) \iff
    \Delta U_2 = W_{1,1} \oplus W_{1,2} \oplus 0 \iff
    \Delta U_2 = \Delta W_1 \quad \blacksquare$$
</ul>

<h4>Anwendung: Extraktion von $K_5$</h4>
<p>Um nicht direkt einmal den kompletten 80-bit Masterkey zu extrahieren, führen wir das gesamte biser besprochene einmal für den ersten iterativen Schritt an $K_5$ durch, das soll dazu dienen, ob man den theoretischen Ablauf verstanden hat, um es auf das gesamte auszuweiten im nachfolgenden Schritt:</p>

<div style="background: #f4f6f9; border-left: 4px solid #17a2b8; padding: 15px; border-radius: 0 4px 4px 0; margin: 15px 0;">
    <strong>Kochrezept</strong>
    <ul style="margin-top: 5px; margin-bottom: 0;">
        1. encryption oracle für plaintextpaare mit der optimalen Eingabedifferenz $\Delta X$ des Trails benutzen<br>
        2. Suchraum durch aktive S-Boxen <i>(mittels bitmask)</i> reduzieren<br>
        3. Für jede mgl. Bit-Kombi der Maske entschlüsseln wir das ciphertextpaar partiell eine Runde rückwärts<br>
        4. Ergibt die Differenz der entschlüsselten Zustände exakt die erwartete Ausgabedifferenz $\Delta Y$ des Trails $\implies$ hitcount++<br>
        $\Rightarrow$ aus den subkey-kandidaten ist der mit den meisten hits <u>statistisch</u> der korrekte <i>subkey</i><br>
    </ul>
</div>

Um das nun durchzuführen, importieren und instanziieren wir unsere "Blackbox".
Zusätzlich benötigen wir itertools für alle mgl. Bit-Kombis.

```python
from modules import EncryptionOracle
oracle = EncryptionOracle() # random 80-bit masterkey
```

```python
target_round_idx = attack.num_rounds - 1

trails_r3 = attack.get_highly_probable_trails(round_num=target_round_idx)
active_trails_r3 = attack.find_trail_combinations(trails_r3)

# masken berechnen und cachen
cached_trails = []
for trail in active_trails_r3:
    mask = attack.get_active_key_mask(trail[3], target_round_idx)
    cached_trails.append((trail, mask))

# darstellen der angewandten trails
table_data = [["trail", "eingabedifferenz", "ausgabedifferenz", "aktive k5-maske"]]
for i, (trail, mask) in enumerate(cached_trails):
    table_data.append([str(i+1), f"0x{trail[2]:04X}", f"0x{trail[3]:04X}", f"0x{mask:04X}"])

print("--- 1. darstellung der angewandten trails ---")
print(table(table_data, header_row=True, frame=True, align='center'))

# extraktion + hit countings
print("\n--- 2. hit-counting pro trail ---")
partial_candidates = []

for i, (trail, mask) in enumerate(cached_trails):
    candidates_with_hits = attack.extract_partial_subkey(
        oracle=oracle.encrypt,
        trail=trail,
        mask=mask,
        target_round=target_round_idx
    )

    pure_candidates = [k for k, _ in candidates_with_hits]
    partial_candidates.append((mask, pure_candidates))


    top_3 = [f"0x{k:04X} ({hits} hits)" for k, hits in candidates_with_hits[:3]]
    print(f"trail {i+1} -> top 3: {top_3}")

# finalresultat mittels merge logik
full_key_hypotheses = attack.merge_key_candidates(partial_candidates)
k5_extracted = full_key_hypotheses[0] if full_key_hypotheses else 0

print("\n--- 3. finales resultat (k5) ---")
print(f"zusammengesetztes k5: 0x{k5_extracted:04X}")
```

Laufzeit hierbei im Durchschnitt +1min

```python
extracted_keys = attack.execute_complete_break(oracle.encrypt, descriptive=True)
```

Abschließend überprüfen wir, dass das hier kein Pfusch war.

```python
verification_results = oracle.verify_keys(extracted_keys)

table_data = [["subkey", "expected [0x]", "extracted [0x]", "expected [0b]", "extracted [0b]", "Status"]]

for res in verification_results:
    status = "perfetto" if res["is_correct"] else "something gone wrong..."
    table_data.append([
        res["subkey"],
        res["expected_hex"],
        res["extracted_hex"],
        res["expected_bin"],
        res["extracted_bin"],
        status
    ])

verification_table = table(table_data, header_row=True, frame=True, align='center')
print("verifikationstabelle:")
print(verification_table)
```
